Praktikum 1 - Bubble Sort

```python
def bubble_sort(data):
  n = len(data)
  for i in range(n):
    for j in range(0, n - i - 1):
      if data[j] > data[j + 1]:
        data[j], data[j + 1] = data[j + 1], data[j]
  return data

angka = [64, 34, 25, 12, 22, 11, 90]
print("Data sebelum diurutkan:", angka)
print("Data setelah Bubble Sort:", bubble_sort(angka))
```

praktikum 2 - Selection Sort

```python
def selection_sort(data):
    n = len(data)
    for i in range(n):
       min_index = 1
       for j in range(i + 1, n):
           if data[j] < data[min_index]:
               min_index = j
       data[i], data[min_index] = data[min_index], data[i]
    return data

angka = [64, 25, 12, 22, 11]
print("Data sebelum diurutkan:", angka)
print("Data setelah Selection Sort:", selection_sort(angka))
```

praktikum 3 - Insertion Sort

```python
def insertion_sort(data):
  n = len(data)
  for i in range(1, n):
    key = data[i]
    j = i - 1
    while j >= 0 and data[j] > key:
      data[j + 1] = data[j]
      j -= 1
    data[j + 1] = key
  return data

angka = [12, 11, 13, 5, 6]
print("Data sebelum diurutkan:", angka)
print("Data setelah Insertion Sort:", insertion_sort(angka))
```

praktikum 4 - Studi Kasus : Mengurutkan Nilai Mahasiswa

```python
def bubble_sort_nilai(data):
  n = len(data)
  for i in range(n):
    for j in range(0, n - i - 1):
      if data[j][1] < data[j + 1][1]:
        data[j], data[j + 1] = data[j + 1], data[j]
  return data

mahasiswa = [("Andi", 80), ("Budi", 70), ("Cici", 90), ("Doni", 85)]
print("Data sebelum diurutkan:", mahasiswa)
print("Data setelah diurutkan (desc):", bubble_sort_nilai(mahasiswa))
```

praktikum 6 - Sorting Data Barang (Nama & Harga)

```python
def bubble_sort_key(data, key_index=0, reverse=False):
  n = len(data)
  for i in range(n):
    for j in range(0, n - i - 1):
      if reverse:
        if data[j][key_index] < data[j + 1][key_index]:
          data[j], data[j + 1] = data[j + 1], data[j]
  return data

barang = [("Buku", 15000), ("Pensil", 5000), ("Tas", 75000), ("Pulpen", 10000)]
print("Urut berdasarkan nama (asc):", bubble_sort_key(barang.copy(), 0))
print("Urut berdasarkan harga (desc):", bubble_sort_key(barang.copy(), reverse=True))
```

penugasan 5 - Analisis Kompleksitas

```python
import random

def bubble_sort_count(data):
    n = len(data)
    count = 0
    for i in range(n):
        for j in range(0, n - i - 1):
            count += 1 # Menghitung setiap kali 'if' dijalankan
            if data[j] > data[j + 1]:
                data[j], data[j + 1] = data[j + 1], data[j]
    return count

def selection_sort_count(data):
    n = len(data)
    count = 0
    for i in range(n):
        min_idx = i
        for j in range(i + 1, n):
            count += 1 # Menghitung perbandingan data
            if data[j] < data[min_idx]:
                min_idx = j
        data[i], data[min_idx] = data[min_idx], data[i]
    return count

def insertion_sort_count(data):
    n = len(data)
    count = 0
    for i in range(1, n):
        key = data[i]
        j = i - 1
        # Loop ini melakukan perbandingan
        while j >= 0:
            count += 1
            if data[j] > key:
                data[j + 1] = data[j]
                j -= 1
            else:
                break
        data[j + 1] = key
    return count

# Test untuk ukuran 10, 50, 100
for size in [10, 50, 100]:
    test_data = [random.randint(1, 1000) for _ in range(size)]
    print(f"Ukuran {size}:")
    print(f"  Bubble Sort   : {bubble_sort_count(test_data.copy())} perbandingan")
    print(f"  Selection Sort: {selection_sort_count(test_data.copy())} perbandingan")
    print(f"  Insertion Sort: {insertion_sort_count(test_data.copy())} perbandingan")
```

penugasan 6 - Proyek Mini

```python
def main():
    daftar_barang = []

    print("--- Program Manajemen Barang ---")

    # 1. Input Data Barang
    while True:
        nama = input("\nMasukkan nama barang (atau ketik 'selesai' untuk lanjut): ")
        if nama.lower() == 'selesai':
            break
        try:
            harga = int(input(f"Masukkan harga untuk {nama}: "))
            daftar_barang.append({"nama": nama, "harga": harga})
        except ValueError:
            print("Error: Harga harus berupa angka!")

    if not daftar_barang:
        print("Daftar barang kosong.")
        return

    # 2. Pilihan Pengurutan
    print("\n--- Opsi Pengurutan ---")
    print("1. Urutkan berdasarkan Nama")
    print("2. Urutkan berdasarkan Harga")
    pilihan_kriteria = input("Pilih kriteria (1/2): ")

    print("\n1. Ascending (A-Z / Terkecil ke Terbesar)")
    print("2. Descending (Z-A / Terbesar ke Terkecil)")
    pilihan_urutan = input("Pilih tipe urutan (1/2): ")

    # Tentukan kunci pengurutan (nama atau harga)
    kunci = "nama" if pilihan_kriteria == "1" else "harga"

    # Tentukan arah pengurutan
    is_reverse = True if pilihan_urutan == "2" else False

    # 3. Proses Pengurutan
    # Menggunakan fungsi lambda untuk mengambil nilai dari dictionary
    daftar_barang.sort(key=lambda x: x[kunci], reverse=is_reverse)

    # 4. Tampilkan Hasil
    print(f"\n--- Daftar Barang (Urut berdasarkan {kunci}) ---")
    print(f"{'No':<4} | {'Nama Barang':<20} | {'Harga':<10}")
    print("-" * 40)
    for i, barang in enumerate(daftar_barang, 1):
        print(f"{i:<4} | {barang['nama']:<20} | Rp{barang['harga']:,}")

if __name__ == "__main__":
    main()
```
