Penugasan 1 : Eksperimen dengan Modul math

```python
import math

def luas_lingkaran(r):
    return math.pi * (r ** 2)

jari1 = 7
jari2 = 10

hasil1 = luas_lingkaran(jari1)
hasil2 = luas_lingkaran(jari2)

print("Luas lingkaran dengan jari-jari", jari1, "=", hasil1)
print("Luas lingkaran dengan jari-jari", jari2, "=", hasil2)
```

Penugasan 4 : Modul Buatan Sendiri

```python
%%writefile konversi.py

def celcius_ke_fahrenheit(celcius):
  return (celcius * 9/5) + 32

def fahrenheit_ke_celcius(fahrenheit):
  return (fahrenheit - 32) * 5/9
```

```python
import konversi

celcius = 25
fahrenheit = 7

hasil1 = konversi.celcius_ke_fahrenheit(celcius)
hasil2 = konversi.fahrenheit_ke_celcius(fahrenheit)

print(celcius, "Celcius =", hasil1, "Fahrenheit")
print(fahrenheit, "Fahrenheit =", hasil2, "Celcius")
```

Penugasan 6 : Studi Kasus Mini

```python
%%writefile kuis.py

def ringkasan(t, b):
    print("\n===== HASIL KUIS =====")
    print(f"Total Soal : {t}")
    print(f"Jawaban Benar : {b}")
    print(f"Jawaban Salah : {t - b}")
    print(f"Skor Akhir : {(b/t)*100}%")
    print("==========================")
```

```python
import random
import kuis

total_soal = 5
benar = 0

for _ in range(total_soal):
    a, b = random.randint(1,20), random.randint(1,20)
    jawab = int(input(f"Soal: {a} + {b} ="))
    if jawab == a + b:
        print("Jawaban kamu benar!\n")
        benar += 1
    else:
        print(f"Salah, jawabannya {a+b}\n")
kuis.ringkasan(total_soal, benar)
```
